## Question 3 — Calculate a Phone Bill

Write a function called `calculate_bill(minutes, rate)` that calculates
the total phone bill based on the following rules:

Rules:
- Total = minutes × rate per minute
- Minimum charge is 15 EGP (if total is less than 15, charge 15 EGP)
- If minutes used are more than 100, apply a 20% discount on the total

Parameters:
- minutes : number of minutes used (**float**)
- rate    : price per minute in EGP (**float**), default = 0.50

Returns:
- The final bill amount (float)

Then ask the user to enter the number of minutes,
call the function, and print the result.

###**Expected Output Example:**
-----------------------------
```
Minutes Used : 120
Rate         : 0.50 EGP/min
Discount     : 20% applied
Total Bill   : 48.00 EGP
```
-----------------------------
```
Minutes Used : 30
Rate         : 0.50 EGP/min
Discount     : None
Total Bill   : 15.00 EGP  ← minimum charge applied
```
-----------------------------


````python
# Question 3 — Calculate a Phone Bill

def calculate_bill(minutes, rate=0.50):
    total = minutes * rate

    # Apply 20% discount if more than 100 minutes
    if minutes > 100:
        total = total * 0.80

    # Minimum charge is 15 EGP
    if total < 15:
        total = 15.0

    return total


minutes = float(input("Enter the number of minutes used: "))
rate = 0.50
bill = calculate_bill(minutes, rate)

discount = "20% applied" if minutes > 100 else "None"

print(f"Minutes Used : {minutes:g}")
print(f"Rate         : {rate:.2f} EGP/min")
print(f"Discount     : {discount}")
print(f"Total Bill   : {bill:.2f} EGP")
````

````python
# Question 4 — Call Center: Subscriber Package Update

subscribers = [
    {"name": "Ahmed Ali",    "phone": "01501000567", "package": "Gold",   "balance": 150.0},
    {"name": "Mona Hassan",  "phone": "01512111111", "package": "Silver", "balance": 30.0},
    {"name": "Karim Sami",   "phone": "01525556789", "package": "Bronze", "balance": 80.0},
    {"name": "Sara Khaled",  "phone": "01534555090", "package": "Gold",   "balance": 250.0},
    {"name": "Omar Mahmoud", "phone": "01545678901", "package": "Silver", "balance": 120.0},
]

prices = {"Bronze": 50, "Silver": 100, "Gold": 200}

# 1) Print all subscribers
print("--- Subscriber List ---")
for sub in subscribers:
    print(f"{sub['name']:<12} | {sub['phone']} | {sub['package']:<6} | Balance: {sub['balance']:.2f} EGP")

# 2) Ask for phone number
phone = input("\nEnter subscriber phone number: ")

# 3) Search for the subscriber
found = None
for sub in subscribers:
    if sub["phone"] == phone:
        found = sub
        break

if found:
    print("Subscriber Found:")
    print(f"Name    : {found['name']}")
    print(f"Phone   : {found['phone']}")
    print(f"Package : {found['package']}")
    print(f"Balance : {found['balance']:.2f} EGP")

    # 4) Ask for the new package
    new_package = input("Enter new package (Bronze / Silver / Gold): ").capitalize()

    if new_package not in prices:
        print("Invalid package name!")
    else:
        price = prices[new_package]
        # 5) Check balance
        if found["balance"] >= price:
            found["balance"] -= price
            found["package"] = new_package
            print("\u2713 Package updated successfully!")
            print(f"Name    : {found['name']}")
            print(f"Package : {found['package']} (updated)")
            print(f"Balance : {found['balance']:.2f} EGP")
        else:
            needed = price - found["balance"]
            print("\u2717 Insufficient balance!")
            print(f"{new_package} package costs {price} EGP.")
            print(f"Current balance : {found['balance']:.2f} EGP")
            print(f"Please recharge : {needed:.2f} EGP to proceed.")
else:
    # 6) Not found
    print("Subscriber not found. Cannot update.")
````

## Question 5 — Analyze Data Usage

Write a **function** called `analyze_usage` that takes a list of data usage values (in GB)  
and returns a dictionary containing:
- `average` — the average usage
- `max` — the highest usage
- `min` — the lowest usage

Then call the function with this list and print the results:
```python
usage = [4.5, 12.0, 7.3, 9.8, 2.1, 15.0, 6.6]
```

**Expected Output Example:**
```
Average Usage : 8.19 GB
Max Usage     : 15.0 GB
Min Usage     : 2.1 GB
```



````python
# Question 5 — Analyze Data Usage

def analyze_usage(usage_list):
    return {
        "average": sum(usage_list) / len(usage_list),
        "max": max(usage_list),
        "min": min(usage_list),
    }


usage = [4.5, 12.0, 7.3, 9.8, 2.1, 15.0, 6.6]
result = analyze_usage(usage)

print(f"Average Usage : {result['average']:.2f} GB")
print(f"Max Usage     : {result['max']} GB")
print(f"Min Usage     : {result['min']} GB")
````

## Question 6 — Save Complaints to a File

Write a program with two functions:

1. `add_complaint(phone, complaint)` — saves the complaint to a file called `complaints.txt`  
   in the format: `phone | complaint`

2. `show_complaints()` — reads and prints all saved complaints from the file.  
   If the file does not exist, print: `No complaints found.`

Ask the user for their phone number and complaint, then call both functions.

**Expected Output Example:**
```
Complaint saved successfully!

--- All Complaints ---
01512111111 | Internet is very slow
01501000567 | Cannot make calls
```


````python
# Question 6 — Save Complaints to a File

import os

def add_complaint(phone, complaint):
    with open("complaints.txt", "a") as f:
        f.write(f"{phone} | {complaint}\n")
    print("Complaint saved successfully!")


def show_complaints():
    if not os.path.exists("complaints.txt"):
        print("No complaints found.")
        return
    print("\n--- All Complaints ---")
    with open("complaints.txt", "r") as f:
        for line in f:
            print(line.strip())


phone = input("Enter your phone number: ")
complaint = input("Enter your complaint: ")

add_complaint(phone, complaint)
show_complaints()
````

## Question 7 — Subscriber Management System (OOP)
---

### Part A — Base Class

Create a class called `Subscriber` with:

**Attributes (`__init__`):**
- `name`, `phone`, `package` (Bronze / Silver / Gold), `balance` (default = 0)

**Package Prices (store inside the class):**
| Package | Monthly Price |
|---------|--------------|
| Bronze  | 50 EGP       |
| Silver  | 100 EGP      |
| Gold    | 200 EGP      |

**Methods:**
- `recharge(amount)` — adds amount to balance, prints new balance.  
  Raise `ValueError` if amount ≤ 0.
- `deduct(amount)` — deducts amount from balance.  
  Raise `ValueError` if amount ≤ 0.  
  Raise `Exception` if balance is insufficient.
- `upgrade_package(new_package)` — changes the package **only if**:
  1. `new_package` is valid (Bronze / Silver / Gold)
  2. Balance is enough to cover the new package price  
  Deduct the price and confirm. If not enough → print how much is needed.
- `show_info()` — prints all subscriber details including balance.
- `__str__()` — returns a formatted string of the subscriber (used when printing the object directly).

---

### Part B

Create two subclasses:

**`PrepaidSubscriber(Subscriber)`:**
- Has an extra attribute: `data_gb` (default = 0)
- Override `recharge(amount)`: after recharging, add 1 GB for every 10 EGP recharged.
- Add method `check_data()`: prints remaining data in GB.

**`PostpaidSubscriber(Subscriber)`:**
- Has an extra attribute: `credit_limit` (default = 500 EGP)
- Override `deduct(amount)`: allow balance to go **negative** but not below `-credit_limit`.  
  If it exceeds the limit → raise `Exception("Credit limit reached!")`.
- Add method `show_debt()`: if balance is negative, print how much the subscriber owes.

---

### Part C

1. Create a list called `subscribers` containing:
   - 2 `PrepaidSubscriber` objects
   - 2 `PostpaidSubscriber` objects
   - 1 `Subscriber` object (base)

2. Print all subscribers using a `for` loop (use `__str__`).

3. Write a function `find_by_phone(subscribers, phone)`:
   - Returns the subscriber object if found.
   - Raises `ValueError("Subscriber not found.")` if not.

4. Write a function `filter_by_package(subscribers, package)`:
   - Returns a list of all subscribers on that package.

5. Write a function `top_balance(subscribers)`:
   - Returns the subscriber with the highest balance.

---

### Part D

Using the list from Part C:
1. Find a subscriber by phone → upgrade their package.
2. Recharge a `PrepaidSubscriber` → check their data.
3. Deduct from a `PostpaidSubscriber` until credit limit is hit → handle the exception.
4. Filter all `Gold` subscribers and print them.
5. Print the subscriber with the top balance.

---
```
**Expected Output Example:**
--- All Subscribers ---

[Prepaid]  Ahmed Ali   | 01501555555 | Bronze | Balance: 0.00 EGP  | Data: 0 GB

[Prepaid]  Mona Hassan | 01512118555 | Silver | Balance: 0.00 EGP  | Data: 0 GB

[Postpaid] Karim Sami  | 01525556789 | Gold   | Balance: 0.00 EGP  | Limit: 500 EGP

[Postpaid] Sara Khaled | 01534555090 | Bronze | Balance: 0.00 EGP  | Limit: 500 EGP

[Base]     Omar Mahmoud| 01645678901 | Silver | Balance: 0.00 EGP
Recharging Ahmed Ali with 100 EGP...

Balance: 100.00 EGP | Data added: 10 GB
Upgrading Ahmed Ali to Gold...

✓ Package updated! Balance: 0.00 EGP (200 EGP deducted)
Deducting from Karim Sami...

✗ Credit limit reached!
--- Gold Subscribers ---

Karim Sami | 01223456789 | Gold
Top Balance: Omar Mahmoud | 50.00 EGP

```

````python
# Question 7 — Subscriber Management System (OOP)

# ---------- Part A: Base Class ----------
class Subscriber:
    PACKAGE_PRICES = {"Bronze": 50, "Silver": 100, "Gold": 200}

    def __init__(self, name, phone, package, balance=0):
        self.name = name
        self.phone = phone
        self.package = package
        self.balance = balance

    def recharge(self, amount):
        if amount <= 0:
            raise ValueError("Recharge amount must be positive.")
        self.balance += amount
        print(f"Balance: {self.balance:.2f} EGP")

    def deduct(self, amount):
        if amount <= 0:
            raise ValueError("Deduct amount must be positive.")
        if amount > self.balance:
            raise Exception("Insufficient balance!")
        self.balance -= amount

    def upgrade_package(self, new_package):
        if new_package not in self.PACKAGE_PRICES:
            print("Invalid package name!")
            return
        price = self.PACKAGE_PRICES[new_package]
        if self.balance >= price:
            self.balance -= price
            self.package = new_package
            print(f"\u2713 Package updated! Balance: {self.balance:.2f} EGP ({price} EGP deducted)")
        else:
            needed = price - self.balance
            print(f"\u2717 Not enough balance. Please recharge {needed:.2f} EGP.")

    def show_info(self):
        print(f"Name    : {self.name}")
        print(f"Phone   : {self.phone}")
        print(f"Package : {self.package}")
        print(f"Balance : {self.balance:.2f} EGP")

    def __str__(self):
        return f"[Base]     {self.name:<12}| {self.phone} | {self.package:<6} | Balance: {self.balance:.2f} EGP"


# ---------- Part B: Subclasses ----------
class PrepaidSubscriber(Subscriber):
    def __init__(self, name, phone, package, balance=0, data_gb=0):
        super().__init__(name, phone, package, balance)
        self.data_gb = data_gb

    def recharge(self, amount):
        if amount <= 0:
            raise ValueError("Recharge amount must be positive.")
        self.balance += amount
        bonus_data = int(amount // 10)  # 1 GB for every 10 EGP
        self.data_gb += bonus_data
        print(f"Balance: {self.balance:.2f} EGP | Data added: {bonus_data} GB")

    def check_data(self):
        print(f"Remaining data: {self.data_gb} GB")

    def __str__(self):
        return f"[Prepaid]  {self.name:<12}| {self.phone} | {self.package:<6} | Balance: {self.balance:.2f} EGP | Data: {self.data_gb} GB"


class PostpaidSubscriber(Subscriber):
    def __init__(self, name, phone, package, balance=0, credit_limit=500):
        super().__init__(name, phone, package, balance)
        self.credit_limit = credit_limit

    def deduct(self, amount):
        if amount <= 0:
            raise ValueError("Deduct amount must be positive.")
        if self.balance - amount < -self.credit_limit:
            raise Exception("Credit limit reached!")
        self.balance -= amount

    def show_debt(self):
        if self.balance < 0:
            print(f"{self.name} owes {abs(self.balance):.2f} EGP")
        else:
            print(f"{self.name} has no debt.")

    def __str__(self):
        return f"[Postpaid] {self.name:<12}| {self.phone} | {self.package:<6} | Balance: {self.balance:.2f} EGP | Limit: {self.credit_limit} EGP"


# ---------- Part C ----------
subscribers = [
    PrepaidSubscriber("Ahmed Ali", "01501555555", "Bronze"),
    PrepaidSubscriber("Mona Hassan", "01512118555", "Silver"),
    PostpaidSubscriber("Karim Sami", "01525556789", "Gold"),
    PostpaidSubscriber("Sara Khaled", "01534555090", "Bronze"),
    Subscriber("Omar Mahmoud", "01645678901", "Silver", balance=50),
]

print("--- All Subscribers ---")
for sub in subscribers:
    print(sub)


def find_by_phone(subscribers, phone):
    for sub in subscribers:
        if sub.phone == phone:
            return sub
    raise ValueError("Subscriber not found.")


def filter_by_package(subscribers, package):
    return [sub for sub in subscribers if sub.package == package]


def top_balance(subscribers):
    return max(subscribers, key=lambda sub: sub.balance)


# ---------- Part D ----------
# 1) Find a subscriber by phone -> recharge then upgrade their package
print("\nRecharging Ahmed Ali with 250 EGP...")
ahmed = find_by_phone(subscribers, "01501555555")
ahmed.recharge(250)

print("Upgrading Ahmed Ali to Gold...")
ahmed.upgrade_package("Gold")

# 2) Recharge a PrepaidSubscriber -> check their data
print("\nRecharging Mona Hassan with 100 EGP...")
mona = find_by_phone(subscribers, "01512118555")
mona.recharge(100)
mona.check_data()

# 3) Deduct from a PostpaidSubscriber until credit limit is hit
print("\nDeducting from Karim Sami...")
karim = find_by_phone(subscribers, "01525556789")
try:
    while True:
        karim.deduct(200)
except Exception as e:
    print(f"\u2717 {e}")
karim.show_debt()

# 4) Filter all Gold subscribers
print("\n--- Gold Subscribers ---")
for sub in filter_by_package(subscribers, "Gold"):
    print(f"{sub.name} | {sub.phone} | {sub.package}")

# 5) Subscriber with the top balance
top = top_balance(subscribers)
print(f"\nTop Balance: {top.name} | {top.balance:.2f} EGP")
````
